```python
import pandas as pd
import numpy as np
import os

print("Libraries imported successfully!")
```

```python
file_name = list(uploaded.keys())[0]

print("Uploaded file:", file_name)
```

```python
# Read all sheets from the Excel workbook
all_sheets = pd.read_excel(
    file_name,
    sheet_name=None
)

# Display available sheet names
print("Available sheets:")
print(list(all_sheets.keys()))
```

```python
customers = all_sheets["Customers"]
policies = all_sheets["Policies"]
claims = all_sheets["Claims"]
payments = all_sheets["Payments"]
investigations = all_sheets["Investigations"]
claim_types = all_sheets["Claim_Types"]

print("All six data tables loaded successfully!")
```

```python
# Store all six DataFrames in one dictionary
tables = {
    "Customers": customers,
    "Policies": policies,
    "Claims": claims,
    "Payments": payments,
    "Investigations": investigations,
    "Claim_Types": claim_types
}

# Display rows and columns of every table
for table_name, df in tables.items():
    print(
        table_name,
        " | Rows:", df.shape[0],
        " | Columns:", df.shape[1]
    )
```

```python
print("CUSTOMERS")
display(customers.head())

print("POLICIES")
display(policies.head())

print("CLAIMS")
display(claims.head())

print("PAYMENTS")
display(payments.head())

print("INVESTIGATIONS")
display(investigations.head())

print("CLAIM TYPES")
display(claim_types.head())
```

```python
for table_name, df in tables.items():
    print("\n==========", table_name, "==========")
    df.info()
```

```python
for table_name, df in tables.items():
    print("\n", table_name, "columns:")
    print(df.columns.tolist())
```

```python
for table_name, df in tables.items():
    print("\n==========", table_name, "==========")
    print(df.isnull().sum())
```

```python
for table_name, df in tables.items():

    missing_count = df.isnull().sum()

    missing_percentage = (
        df.isnull().mean() * 100
    ).round(2)

    missing_report = pd.DataFrame({
        "Missing Count": missing_count,
        "Missing Percentage": missing_percentage
    })

    print("\n==========", table_name, "==========")
    display(missing_report)
```

```python
for table_name, df in tables.items():

    duplicate_count = df.duplicated().sum()

    print(
        table_name,
        "Exact duplicate rows:",
        duplicate_count
    )
```

```python
primary_keys = {
    "Customers": ("CustomerID", customers),
    "Policies": ("PolicyID", policies),
    "Claims": ("ClaimID", claims),
    "Payments": ("PaymentID", payments),
    "Investigations": ("InvestigationID", investigations),
    "Claim_Types": ("ClaimTypeID", claim_types)
}

for table_name, (key_column, df) in primary_keys.items():

    duplicate_keys = df[key_column].duplicated().sum()

    print(
        table_name,
        "| Duplicate", key_column, ":",
        duplicate_keys
    )
```

```python
invalid_customer_ids = policies[
    ~policies["CustomerID"].isin(customers["CustomerID"])
]

print("Invalid CustomerID records:", len(invalid_customer_ids))
```

```python
invalid_policy_ids = claims[
    ~claims["PolicyID"].isin(policies["PolicyID"])
]

print("Invalid PolicyID records:", len(invalid_policy_ids))
```

```python
invalid_claim_type_ids = claims[
    ~claims["ClaimTypeID"].isin(claim_types["ClaimTypeID"])
]

print("Invalid ClaimTypeID records:", len(invalid_claim_type_ids))
```

```python
invalid_payment_claims = payments[
    ~payments["ClaimID"].isin(claims["ClaimID"])
]

print("Invalid ClaimID in Payments:", len(invalid_payment_claims))
```

```python
invalid_investigation_claims = investigations[
    ~investigations["ClaimID"].isin(claims["ClaimID"])
]

print("Invalid ClaimID in Investigations:", len(invalid_investigation_claims))
```

```python
date_columns = {
    "Customers": ["CustomerSince"],
    "Policies": ["PolicyStartDate", "PolicyEndDate"],
    "Claims": ["IncidentDate", "ReportedDate"],
    "Payments": ["PaymentDate"],
    "Investigations": ["InvestigationDate"]
}

for table_name, columns in date_columns.items():

    df = tables[table_name]

    for column in columns:
        df[column] = pd.to_datetime(
            df[column],
            errors="coerce"
        )

    print(table_name, "date columns converted.")
```

```python
for table_name, columns in date_columns.items():

    print("\n", table_name)

    for column in columns:
        print(
            column,
            "| Missing/Invalid dates:",
            tables[table_name][column].isna().sum()
        )
```

```python
invalid_age = customers[
    (customers["CustomerAge"] < 18) |
    (customers["CustomerAge"] > 100)
]

print("Invalid customer ages:", len(invalid_age))
```

```python
invalid_claim_amount = claims[
    claims["ClaimAmountINR"] <= 0
]

print("Invalid claim amounts:", len(invalid_claim_amount))
```

```python
invalid_payment_amount = payments[
    payments["PaymentAmountINR"] < 0
]

print("Negative payment amounts:", len(invalid_payment_amount))
```

```python
invalid_claim_dates = claims[
    claims["ReportedDate"] < claims["IncidentDate"]
]

print("Reported before incident:", len(invalid_claim_dates))
```

```python
invalid_policy_dates = policies[
    policies["PolicyEndDate"] < policies["PolicyStartDate"]
]

print("Invalid policy date ranges:", len(invalid_policy_dates))

print(claims["FraudFlag"].value_counts(dropna=False))
```

```python
for table_name, df in tables.items():

    print("\n==========", table_name, "==========")

    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Missing values:", df.isnull().sum().sum())
    print("Exact duplicate rows:", df.duplicated().sum())
```

```python
output_file = "Insurance_Claims_Intelligence_Cleaned.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

    # Save six processed data tables
    for table_name, df in tables.items():
        df.to_excel(
            writer,
            sheet_name=table_name,
            index=False
        )

    # Preserve the original README sheet
    if "README" in all_sheets:
        all_sheets["README"].to_excel(
            writer,
            sheet_name="README",
            index=False
        )

print("Cleaned workbook saved successfully!")
```
